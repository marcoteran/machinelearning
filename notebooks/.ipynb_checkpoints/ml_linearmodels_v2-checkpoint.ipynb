## Introducción y objetivos

En esta sesión el curso entra formalmente al **modelado supervisado**.

La familia lineal no aparece aquí como una reliquia histórica ni como un trámite previo a modelos “más potentes”. Aparece como la primera familia seria que nos permite ver con claridad:

- cómo se construye un **score** a partir de features,

### Helpers

Estos helpers buscan tres cosas:

- mantener el notebook legible,
- centralizar la evaluación,
- y evitar errores de compatibilidad o repetición innecesaria.


```python
def print_section(title: str, width: int = 88) -> None:
    print("\n" + "=" * width)
    print(title.center(width))
    print("=" * width)

def rmse_score(y_true, y_pred) -> float:
    try:
        from sklearn.metrics import root_mean_squared_error
        return float(root_mean_squared_error(y_true, y_pred))
    except Exception:
        return float(np.sqrt(mean_squared_error(y_true, y_pred)))

def regression_metrics(y_true, y_pred) -> Dict[str, float]:
    return {
        "rmse": rmse_score(y_true, y_pred),
        "mae": float(mean_absolute_error(y_true, y_pred)),
        "r2": float(r2_score(y_true, y_pred)),
    }

def classification_metrics(y_true, proba, threshold: float, cost_fn: float = 5.0, cost_fp: float = 1.0) -> Dict[str, float]:
    y_pred = (proba >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()
    accuracy = (tp + tn) / (tp + tn + fp + fn)
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0.0
    f1 = (2 * precision * recall / (precision + recall)) if (precision + recall) > 0 else 0.0
    auc = float(roc_auc_score(y_true, proba))
    ll = float(log_loss(y_true, proba))
    policy_cost = float(cost_fn * fn + cost_fp * fp)
    return {
        "threshold": float(threshold),
        "accuracy": float(accuracy),
        "precision": float(precision),
        "recall": float(recall),
        "specificity": float(specificity),
        "f1": float(f1),
        "roc_auc": auc,
        "log_loss": ll,
        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn),
        "tp": int(tp),
        "policy_cost": policy_cost,
    }

def split_train_val_test(X: pd.DataFrame, y: pd.Series, test_size: float = 0.20, val_size_within_train: float = 0.25, stratify: Optional[pd.Series] = None, random_state: int = SEED) -> dict:
    X_train_full, X_test, y_train_full, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state, stratify=stratify)
    stratify_train = y_train_full if stratify is not None else None
    X_train, X_val, y_train, y_val = train_test_split(X_train_full, y_train_full, test_size=val_size_within_train, random_state=random_state, stratify=stratify_train)
    return {"X_train": X_train, "X_val": X_val, "X_test": X_test, "y_train": y_train, "y_val": y_val, "y_test": y_test}

def evaluate_regression_model(model, X_train, y_train, X_val, y_val, X_test, y_test, model_name: str) -> Dict[str, object]:
    fitted = clone(model)
    fitted.fit(X_train, y_train)
    val_pred = fitted.predict(X_val)
    test_pred = fitted.predict(X_test)
    val_metrics = regression_metrics(y_val, val_pred)
    test_metrics = regression_metrics(y_test, test_pred)
    return {
        "model": model_name,
        "val_rmse": val_metrics["rmse"],
        "val_mae": val_metrics["mae"],
        "val_r2": val_metrics["r2"],
        "test_rmse": test_metrics["rmse"],
        "test_mae": test_metrics["mae"],
        "test_r2": test_metrics["r2"],
        "fitted_model": fitted,
    }

def threshold_sweep(y_true, proba, thresholds, cost_fn: float = 5.0, cost_fp: float = 1.0) -> pd.DataFrame:
    return pd.DataFrame([classification_metrics(y_true, proba, float(t), cost_fn=cost_fn, cost_fp=cost_fp) for t in thresholds])

def extract_coefficients(model, feature_names: List[str]) -> pd.DataFrame:
    if isinstance(model, Pipeline):
        final_estimator = model.named_steps[list(model.named_steps.keys())[-1]]
    else:
        final_estimator = model
    return pd.DataFrame({"feature": feature_names, "coef": np.ravel(final_estimator.coef_)})

def normal_equation_simple(x: np.ndarray, y: np.ndarray) -> Tuple[float, float]:
    x = np.asarray(x).reshape(-1)
    y = np.asarray(y).reshape(-1)
    x_mean = x.mean(); y_mean = y.mean()
    beta_1 = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
    beta_0 = y_mean - beta_1 * x_mean
    return float(beta_0), float(beta_1)

def gradient_descent_simple_linear(x: np.ndarray, y: np.ndarray, lr: float = 0.1, n_iters: int = 500, random_state: int = SEED) -> Tuple[float, float, List[float]]:
    rng = np.random.default_rng(random_state)
    x = np.asarray(x).reshape(-1); y = np.asarray(y).reshape(-1)
    beta_0 = rng.normal(); beta_1 = rng.normal(); history = []; n = len(x)
    for _ in range(n_iters):
        y_hat = beta_0 + beta_1 * x
        error = y_hat - y
        history.append(float(np.mean(error ** 2)))
        grad_b0 = (2 / n) * np.sum(error)
        grad_b1 = (2 / n) * np.sum(error * x)
        beta_0 -= lr * grad_b0
        beta_1 -= lr * grad_b1
    return float(beta_0), float(beta_1), history

def sigmoid(z: np.ndarray) -> np.ndarray:
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))
```

## Regresión lineal: intuición visual

Para entender la regresión lineal conviene empezar con una sola variable.  
Usaremos `bmi` del dataset de diabetes.

Primero veremos:

- nube de puntos,
- recta ajustada,
- residuos,
- y después pasaremos a tres formas de resolver el mismo problema:
  1. fórmula cerrada,
  2. gradient descent,
  3. librería estándar.


```python
x_bmi = X_reg_full["bmi"].values
y_reg_np = y_reg_full.values

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].scatter(x_bmi, y_reg_np, alpha=0.7)
axes[0].set_title("Diabetes: BMI vs progresión de enfermedad")
axes[0].set_xlabel("bmi")
axes[0].set_ylabel("target")

axes[1].hist(y_reg_np, bins=30, edgecolor="black", alpha=0.8)
axes[1].set_title("Distribución del target de regresión")
axes[1].set_xlabel("target")
axes[1].set_ylabel("frecuencia")

plt.tight_layout(); plt.show()
```

### Idea central del bloque

En regresión lineal simple queremos ajustar:

\[
\hat{y} = eta_0 + eta_1 x
\]

y minimizar:

\[
MSE = 
rac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
\]

La pregunta pedagógica aquí es importante:

> ¿Qué cambia si resolvemos el mismo problema con fórmula cerrada, con gradient descent o con una librería?


```python
X_simple = pd.DataFrame({"bmi": x_bmi})
simple_split = split_train_val_test(X_simple, y_reg_full, stratify=None, random_state=SEED)

x_train_simple = simple_split["X_train"]["bmi"].values
x_test_simple = simple_split["X_test"]["bmi"].values
y_train_simple = simple_split["y_train"].values
y_test_simple = simple_split["y_test"].values

print_section("SPLIT SIMPLE PARA BMI")
print(f"Train: {len(x_train_simple)}")
print(f"Validation: {len(simple_split['X_val'])}")
print(f"Test: {len(x_test_simple)}")
```

---
## Regresión lineal: solución cerrada y gradient descent

### 1. Fórmula cerrada

Para regresión lineal simple, la solución determinística puede obtenerse sin iterar:

\[
eta_1 = 
rac{\sum (x_i - ar{x})(y_i - ar{y})}{\sum (x_i - ar{x})^2}
\]

\[
eta_0 = ar{y} - eta_1 ar{x}
\]

### 2. Gradient descent

También podemos minimizar la pérdida iterativamente.

La idea no es convertir esta sesión en una clase de optimización, sino mostrar que:

- existe una solución cerrada,
- existe una solución iterativa,
- y ambas deben llevarnos a una respuesta muy parecida si todo está bien configurado.


```python
beta0_closed, beta1_closed = normal_equation_simple(x_train_simple, y_train_simple)
y_pred_closed = beta0_closed + beta1_closed * x_test_simple
closed_metrics = regression_metrics(y_test_simple, y_pred_closed)

print_section("SOLUCIÓN CERRADA — REGRESIÓN LINEAL SIMPLE")
print(f"beta_0 = {beta0_closed:.4f}")
print(f"beta_1 = {beta1_closed:.4f}")
print(closed_metrics)
```

```python
x_mean = x_train_simple.mean(); x_std = x_train_simple.std()
x_train_simple_std = (x_train_simple - x_mean) / x_std
x_test_simple_std = (x_test_simple - x_mean) / x_std

gd_b0, gd_b1, gd_history = gradient_descent_simple_linear(x_train_simple_std, y_train_simple, lr=0.05, n_iters=1200, random_state=SEED)
y_pred_gd = gd_b0 + gd_b1 * x_test_simple_std
gd_metrics = regression_metrics(y_test_simple, y_pred_gd)

print_section("GRADIENT DESCENT — REGRESIÓN LINEAL SIMPLE")
print(f"beta_0 = {gd_b0:.4f}")
print(f"beta_1 = {gd_b1:.4f}")
print(gd_metrics)
```

```python
lin_simple = LinearRegression()
lin_simple.fit(simple_split["X_train"], simple_split["y_train"])
y_pred_sklearn = lin_simple.predict(simple_split["X_test"])
sklearn_metrics_simple = regression_metrics(simple_split["y_test"], y_pred_sklearn)

print_section("SCIKIT-LEARN — REGRESIÓN LINEAL SIMPLE")
print(f"intercept_ = {lin_simple.intercept_:.4f}")
print(f"coef_ = {lin_simple.coef_[0]:.4f}")
print(sklearn_metrics_simple)
```

```python
simple_comparison = pd.DataFrame([
    {"method": "closed_form", **closed_metrics},
    {"method": "gradient_descent", **gd_metrics},
    {"method": "scikit_learn", **sklearn_metrics_simple},
])
print_section("COMPARACIÓN DE LOS TRES MÉTODOS")
display(simple_comparison)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].scatter(x_train_simple, y_train_simple, alpha=0.5, label="train")
axes[0].scatter(x_test_simple, y_test_simple, alpha=0.5, label="test")
x_line = np.linspace(x_train_simple.min(), x_train_simple.max(), 200)
y_line_closed = beta0_closed + beta1_closed * x_line
axes[0].plot(x_line, y_line_closed, linewidth=2, label="solución cerrada")
axes[0].set_title("Recta ajustada sobre BMI")
axes[0].set_xlabel("bmi")
axes[0].set_ylabel("target")
axes[0].legend()

axes[1].plot(gd_history)
axes[1].set_title("Convergencia de gradient descent")
axes[1].set_xlabel("iteración")
axes[1].set_ylabel("MSE train")
plt.tight_layout(); plt.show()
```

```python
residuals_closed = y_test_simple - y_pred_closed
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].scatter(y_test_simple, y_pred_closed, alpha=0.7)
lims = [min(y_test_simple.min(), y_pred_closed.min()), max(y_test_simple.max(), y_pred_closed.max())]
axes[0].plot(lims, lims, linestyle="--")
axes[0].set_title("Observado vs predicho")
axes[0].set_xlabel("target real")
axes[0].set_ylabel("target predicho")

axes[1].scatter(y_pred_closed, residuals_closed, alpha=0.7)
axes[1].axhline(0, linestyle="--")
axes[1].set_title("Residuos en test")
axes[1].set_xlabel("predicción")
axes[1].set_ylabel("residuo")
plt.tight_layout(); plt.show()
```

### Lectura guiada

1. La **fórmula cerrada** y **scikit-learn** deberían dar prácticamente la misma solución para el problema simple.
2. **Gradient descent** no cambia el objetivo; cambia la forma de llegar a la solución.
3. Si la historia de pérdida baja y se estabiliza, tenemos evidencia de convergencia razonable.

## Regresión multivariable: sin preprocessing vs con preprocessing

Ahora sí pasamos al problema más realista:

- múltiples features,
- comparación contra baseline ingenua,
- modelo lineal sin preprocessing,
- modelo lineal con scaling explícito,
- y después regularización.


```python
reg_split = split_train_val_test(X_reg_full, y_reg_full, stratify=None, random_state=SEED)
print_section("SPLIT DE REGRESIÓN")
print(f"Train: {reg_split['X_train'].shape[0]}")
print(f"Validation: {reg_split['X_val'].shape[0]}")
print(f"Test: {reg_split['X_test'].shape[0]}")
```

```python
dummy_reg = DummyRegressor(strategy="mean")
dummy_reg_row = evaluate_regression_model(dummy_reg, reg_split["X_train"], reg_split["y_train"], reg_split["X_val"], reg_split["y_val"], reg_split["X_test"], reg_split["y_test"], "DummyRegressor_mean")
display(pd.DataFrame([{k: v for k, v in dummy_reg_row.items() if k != "fitted_model"}]))
```

```python
linreg_raw = LinearRegression()
linreg_raw_row = evaluate_regression_model(linreg_raw, reg_split["X_train"], reg_split["y_train"], reg_split["X_val"], reg_split["y_val"], reg_split["X_test"], reg_split["y_test"], "LinearRegression_raw")
linreg_raw_fitted = linreg_raw_row["fitted_model"]
display(pd.DataFrame([{k: v for k, v in linreg_raw_row.items() if k != "fitted_model"}]))
```

```python
linreg_scaled = Pipeline([("scaler", StandardScaler()), ("model", LinearRegression())])
linreg_scaled_row = evaluate_regression_model(linreg_scaled, reg_split["X_train"], reg_split["y_train"], reg_split["X_val"], reg_split["y_val"], reg_split["X_test"], reg_split["y_test"], "LinearRegression_scaled")
linreg_scaled_fitted = linreg_scaled_row["fitted_model"]
display(pd.DataFrame([{k: v for k, v in linreg_scaled_row.items() if k != "fitted_model"}]))
```

```python
regression_baseline_comparison = pd.DataFrame([
    {k: v for k, v in dummy_reg_row.items() if k != "fitted_model"},
    {k: v for k, v in linreg_raw_row.items() if k != "fitted_model"},
    {k: v for k, v in linreg_scaled_row.items() if k != "fitted_model"},
]).sort_values("val_rmse")
print_section("BASELINE VS LINEAR REGRESSION")
display(regression_baseline_comparison)
```

```python
coef_raw_df = extract_coefficients(linreg_raw_fitted, list(reg_split["X_train"].columns))
coef_raw_df["abs_raw"] = coef_raw_df["coef"].abs()
coef_scaled_df = extract_coefficients(linreg_scaled_fitted, list(reg_split["X_train"].columns))
coef_scaled_df["abs_scaled"] = coef_scaled_df["coef"].abs()
coef_compare_df = coef_raw_df[["feature", "coef", "abs_raw"]].rename(columns={"coef": "coef_raw"}).merge(coef_scaled_df[["feature", "coef", "abs_scaled"]].rename(columns={"coef": "coef_scaled"}), on="feature", how="left")
display(coef_compare_df.sort_values("abs_scaled", ascending=False))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
plot_raw = coef_compare_df.sort_values("coef_raw")
axes[0].barh(plot_raw["feature"], plot_raw["coef_raw"])
axes[0].set_title("Coeficientes sin preprocessing")
axes[0].set_xlabel("coeficiente")
plot_scaled = coef_compare_df.sort_values("coef_scaled")
axes[1].barh(plot_scaled["feature"], plot_scaled["coef_scaled"])
axes[1].set_title("Coeficientes con StandardScaler")
axes[1].set_xlabel("coeficiente estandarizado")
plt.tight_layout(); plt.show()
```

### Regla de interpretación

- Un coeficiente lineal se lee como cambio esperado en la predicción por una unidad de cambio en la feature, **manteniendo las demás constantes en el sentido del modelo**.
- La magnitud de los coeficientes crudos no siempre es comparable.
- Estandarizar ayuda a comparar mejor el peso relativo dentro del modelo.
- Nada de esto autoriza una conclusión causal por sí sola.

## Regularización: Ridge y Lasso

Ahora añadimos dos ideas centrales del capítulo:

- **Ridge (L2)**: encoge coeficientes sin necesariamente apagarlos;
- **Lasso (L1)**: puede empujar algunos coeficientes exactamente a cero.


```python
alphas = np.logspace(-3, 2, 12)
regression_rows = [dummy_reg_row, linreg_raw_row, linreg_scaled_row]
ridge_path_rows = []; lasso_path_rows = []
best_ridge = None; best_ridge_val_rmse = np.inf
for alpha in alphas:
    ridge_model = Pipeline([("scaler", StandardScaler()), ("model", Ridge(alpha=alpha))])
    row = evaluate_regression_model(ridge_model, reg_split["X_train"], reg_split["y_train"], reg_split["X_val"], reg_split["y_val"], reg_split["X_test"], reg_split["y_test"], f"Ridge_alpha={alpha:.4f}")
    regression_rows.append(row)
    coef_tmp = extract_coefficients(row["fitted_model"], list(reg_split["X_train"].columns))
    coef_tmp["alpha"] = alpha; coef_tmp["model"] = "Ridge"; ridge_path_rows.append(coef_tmp)
    if row["val_rmse"] < best_ridge_val_rmse:
        best_ridge_val_rmse = row["val_rmse"]; best_ridge = row
best_lasso = None; best_lasso_val_rmse = np.inf
for alpha in alphas:
    lasso_model = Pipeline([("scaler", StandardScaler()), ("model", Lasso(alpha=alpha, max_iter=30000))])
    row = evaluate_regression_model(lasso_model, reg_split["X_train"], reg_split["y_train"], reg_split["X_val"], reg_split["y_val"], reg_split["X_test"], reg_split["y_test"], f"Lasso_alpha={alpha:.4f}")
    regression_rows.append(row)
    coef_tmp = extract_coefficients(row["fitted_model"], list(reg_split["X_train"].columns))
    coef_tmp["alpha"] = alpha; coef_tmp["model"] = "Lasso"; lasso_path_rows.append(coef_tmp)
    if row["val_rmse"] < best_lasso_val_rmse:
        best_lasso_val_rmse = row["val_rmse"]; best_lasso = row
regression_results_df = pd.DataFrame([{k: v for k, v in row.items() if k != "fitted_model"} for row in regression_rows]).sort_values("val_rmse")
display(regression_results_df.head(15))
```

```python
ridge_path_df = pd.concat(ridge_path_rows, ignore_index=True)
lasso_path_df = pd.concat(lasso_path_rows, ignore_index=True)
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)
for feature in reg_split["X_train"].columns:
    tmp = ridge_path_df[ridge_path_df["feature"] == feature]
    axes[0].plot(tmp["alpha"], tmp["coef"], label=feature)
axes[0].set_xscale("log")
axes[0].set_title("Trayectoria de coeficientes — Ridge")
axes[0].set_xlabel("alpha")
axes[0].set_ylabel("coeficiente")
for feature in reg_split["X_train"].columns:
    tmp = lasso_path_df[lasso_path_df["feature"] == feature]
    axes[1].plot(tmp["alpha"], tmp["coef"], label=feature)
axes[1].set_xscale("log")
axes[1].set_title("Trayectoria de coeficientes — Lasso")
axes[1].set_xlabel("alpha")
axes[1].legend(loc="best", fontsize=9)
plt.tight_layout(); plt.show()
```

```python
best_regression_models_df = pd.DataFrame([
    {k: v for k, v in dummy_reg_row.items() if k != "fitted_model"},
    {k: v for k, v in linreg_raw_row.items() if k != "fitted_model"},
    {k: v for k, v in linreg_scaled_row.items() if k != "fitted_model"},
    {k: v for k, v in best_ridge.items() if k != "fitted_model"},
    {k: v for k, v in best_lasso.items() if k != "fitted_model"},
]).sort_values("val_rmse")
print_section("SÍNTESIS DE REGRESIÓN")
display(best_regression_models_df)
```

## Regresión logística: probabilidades, log loss y coeficientes

Ahora pasamos a clasificación binaria con el dataset de cáncer.

La idea compartida sigue siendo la del score lineal:

\[
z(x) = eta_0 + \sum_{j=1}^{p} eta_j x_j
\]

Pero ahora el score se transforma en probabilidad con la sigmoide:

\[
\hat{p} = \sigma(z) = 
rac{1}{1 + e^{-z}}
\]


```python
clf_split = split_train_val_test(X_clf_full, y_clf_full, stratify=y_clf_full, random_state=SEED)
print_section("SPLIT DE CLASIFICACIÓN")
print(f"Train: {clf_split['X_train'].shape[0]}")
print(f"Validation: {clf_split['X_val'].shape[0]}")
print(f"Test: {clf_split['X_test'].shape[0]}")
```

```python
dummy_clf = DummyClassifier(strategy="prior", random_state=SEED)
dummy_clf.fit(clf_split["X_train"], clf_split["y_train"])
dummy_val_proba = dummy_clf.predict_proba(clf_split["X_val"])[:, 1]
dummy_test_proba = dummy_clf.predict_proba(clf_split["X_test"])[:, 1]
dummy_clf_val_metrics = classification_metrics(clf_split["y_val"], dummy_val_proba, threshold=0.50)
dummy_clf_test_metrics = classification_metrics(clf_split["y_test"], dummy_test_proba, threshold=0.50)
display(pd.DataFrame([{
    "model": "DummyClassifier_prior",
    "val_log_loss": dummy_clf_val_metrics["log_loss"],
    "val_roc_auc": dummy_clf_val_metrics["roc_auc"],
    "test_log_loss": dummy_clf_test_metrics["log_loss"],
    "test_roc_auc": dummy_clf_test_metrics["roc_auc"],
}]))
```

```python
logreg_raw = LogisticRegression(C=1.0, penalty="l2", solver="liblinear", max_iter=5000, random_state=SEED)
logreg_raw.fit(clf_split["X_train"], clf_split["y_train"])
val_proba_raw = logreg_raw.predict_proba(clf_split["X_val"])[:, 1]
test_proba_raw = logreg_raw.predict_proba(clf_split["X_test"])[:, 1]
raw_metrics_050 = classification_metrics(clf_split["y_test"], test_proba_raw, threshold=0.50)
display(pd.DataFrame([{
    "model": "LogisticRegression_raw",
    "val_log_loss": log_loss(clf_split["y_val"], val_proba_raw),
    "val_roc_auc": roc_auc_score(clf_split["y_val"], val_proba_raw),
    "test_log_loss": log_loss(clf_split["y_test"], test_proba_raw),
    "test_roc_auc": roc_auc_score(clf_split["y_test"], test_proba_raw),
    "test_accuracy_tau_0.5": raw_metrics_050["accuracy"],
    "test_recall_tau_0.5": raw_metrics_050["recall"],
}]))
```

```python
C_grid = np.logspace(-3, 3, 13)
logistic_rows = []
for C in C_grid:
    logistic_model = Pipeline([("scaler", StandardScaler()), ("model", LogisticRegression(C=C, penalty="l2", solver="liblinear", max_iter=5000, random_state=SEED))])
    fitted = clone(logistic_model)
    fitted.fit(clf_split["X_train"], clf_split["y_train"])
    val_proba = fitted.predict_proba(clf_split["X_val"])[:, 1]
    test_proba = fitted.predict_proba(clf_split["X_test"])[:, 1]
    logistic_rows.append({
        "C": C,
        "val_log_loss": log_loss(clf_split["y_val"], val_proba),
        "val_roc_auc": roc_auc_score(clf_split["y_val"], val_proba),
        "test_log_loss": log_loss(clf_split["y_test"], test_proba),
        "test_roc_auc": roc_auc_score(clf_split["y_test"], test_proba),
        "fitted_model": fitted,
    })
logistic_results_df = pd.DataFrame([{k: v for k, v in row.items() if k != "fitted_model"} for row in logistic_rows]).sort_values("val_log_loss")
best_logistic_row = min(logistic_rows, key=lambda row: row["val_log_loss"])
best_logistic_model = best_logistic_row["fitted_model"]
best_C = best_logistic_row["C"]
print_section("RESULTADOS DE LOGÍSTICA CON SCALING")
display(logistic_results_df)
print(f"Mejor C por validation log loss: {best_C:.4f}")
```

```python
val_proba_best = best_logistic_model.predict_proba(clf_split["X_val"])[:, 1]
test_proba_best = best_logistic_model.predict_proba(clf_split["X_test"])[:, 1]
clf_baseline_vs_logistic = pd.DataFrame([
    {"model": "DummyClassifier_prior", "val_log_loss": dummy_clf_val_metrics["log_loss"], "val_roc_auc": dummy_clf_val_metrics["roc_auc"], "test_log_loss": dummy_clf_test_metrics["log_loss"], "test_roc_auc": dummy_clf_test_metrics["roc_auc"]},
    {"model": "LogisticRegression_raw", "val_log_loss": log_loss(clf_split["y_val"], val_proba_raw), "val_roc_auc": roc_auc_score(clf_split["y_val"], val_proba_raw), "test_log_loss": log_loss(clf_split["y_test"], test_proba_raw), "test_roc_auc": roc_auc_score(clf_split["y_test"], test_proba_raw)},
    {"model": f"LogisticRegression_scaled_C={best_C:.4f}", "val_log_loss": log_loss(clf_split["y_val"], val_proba_best), "val_roc_auc": roc_auc_score(clf_split["y_val"], val_proba_best), "test_log_loss": log_loss(clf_split["y_test"], test_proba_best), "test_roc_auc": roc_auc_score(clf_split["y_test"], test_proba_best)},
])
print_section("BASELINE VS LOGÍSTICA")
display(clf_baseline_vs_logistic)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
z = np.linspace(-8, 8, 500)
axes[0].plot(z, sigmoid(z))
axes[0].axhline(0.5, linestyle="--")
axes[0].axvline(0, linestyle="--")
axes[0].set_title("Sigmoide: del score lineal a la probabilidad")
axes[0].set_xlabel("z")
axes[0].set_ylabel("sigmoid(z)")

axes[1].hist(val_proba_best[clf_split["y_val"] == 0], bins=20, alpha=0.6, label="benigno (0)")
axes[1].hist(val_proba_best[clf_split["y_val"] == 1], bins=20, alpha=0.6, label="maligno (1)")
axes[1].set_title("Distribución de probabilidades en validación")
axes[1].set_xlabel("probabilidad estimada de malignidad")
axes[1].set_ylabel("frecuencia")
axes[1].legend()
plt.tight_layout(); plt.show()
```

```python
coef_logistic_df = extract_coefficients(best_logistic_model, list(clf_split["X_train"].columns))
coef_logistic_df["abs_coef"] = coef_logistic_df["coef"].abs()
coef_logistic_df["odds_ratio_aprox_por_+1sd"] = np.exp(coef_logistic_df["coef"])
coef_logistic_df = coef_logistic_df.sort_values("abs_coef", ascending=False).reset_index(drop=True)
display(coef_logistic_df.head(15))
```

## Threshold tuning y métricas según riesgo

Hasta aquí el modelo entrega probabilidades.  
Ahora necesitamos una política de decisión.

Supongamos este escenario:

- la clase positiva es `maligno = 1`,
- un **falso negativo** es más costoso que un falso positivo,
- por tanto, `0.5` no tiene por qué ser el mejor threshold.

Usaremos:

- costo FN = 5
- costo FP = 1


```python
cost_fn = 5.0
cost_fp = 1.0
thresholds = np.linspace(0.05, 0.95, 91)
threshold_df = threshold_sweep(clf_split["y_val"], val_proba_best, thresholds, cost_fn=cost_fn, cost_fp=cost_fp)
best_threshold_row = threshold_df.sort_values(["policy_cost", "fn", "threshold"], ascending=[True, True, True]).iloc[0]
best_threshold = float(best_threshold_row["threshold"])
print_section("THRESHOLD TUNING EN VALIDACIÓN")
display(threshold_df.head())
print(f"Threshold elegido: {best_threshold:.2f}")
print(f"Costo mínimo en validación: {best_threshold_row['policy_cost']:.2f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(threshold_df["threshold"], threshold_df["recall"], label="recall")
axes[0].plot(threshold_df["threshold"], threshold_df["precision"], label="precision")
axes[0].plot(threshold_df["threshold"], threshold_df["specificity"], label="specificity")
axes[0].axvline(0.50, linestyle="--", label="threshold=0.50")
axes[0].axvline(best_threshold, linestyle="--", label=f"threshold óptimo={best_threshold:.2f}")
axes[0].set_title("Métricas según threshold (validación)")
axes[0].set_xlabel("threshold")
axes[0].set_ylabel("valor")
axes[0].legend()

axes[1].plot(threshold_df["threshold"], threshold_df["policy_cost"], label="policy_cost")
axes[1].axvline(0.50, linestyle="--", label="threshold=0.50")
axes[1].axvline(best_threshold, linestyle="--", label=f"threshold óptimo={best_threshold:.2f}")
axes[1].set_title("Costo de política según threshold")
axes[1].set_xlabel("threshold")
axes[1].set_ylabel("costo")
axes[1].legend()
plt.tight_layout(); plt.show()
```

```python
test_default_metrics = classification_metrics(clf_split["y_test"], test_proba_best, threshold=0.50, cost_fn=cost_fn, cost_fp=cost_fp)
test_tuned_metrics = classification_metrics(clf_split["y_test"], test_proba_best, threshold=best_threshold, cost_fn=cost_fn, cost_fp=cost_fp)
classification_summary_df = pd.DataFrame([
    {"policy": "default_0.50", **test_default_metrics},
    {"policy": "tuned_validation_threshold", **test_tuned_metrics},
])
print_section("COMPARACIÓN EN TEST: 0.50 VS THRESHOLD AJUSTADO")
display(classification_summary_df)
```

```python
fpr_raw, tpr_raw, _ = roc_curve(clf_split["y_val"], val_proba_raw)
fpr_best, tpr_best, _ = roc_curve(clf_split["y_val"], val_proba_best)
precision_best, recall_best, _ = precision_recall_curve(clf_split["y_val"], val_proba_best)
ap_best = average_precision_score(clf_split["y_val"], val_proba_best)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(fpr_raw, tpr_raw, label="logistic raw")
axes[0].plot(fpr_best, tpr_best, label="logistic scaled")
axes[0].plot([0, 1], [0, 1], linestyle="--")
axes[0].set_title("ROC en validación")
axes[0].set_xlabel("False Positive Rate")
axes[0].set_ylabel("True Positive Rate")
axes[0].legend()
axes[1].plot(recall_best, precision_best, label=f"scaled logistic (AP={ap_best:.3f})")
axes[1].set_title("Precision-Recall en validación")
axes[1].set_xlabel("Recall")
axes[1].set_ylabel("Precision")
axes[1].legend()
plt.tight_layout(); plt.show()
```

## Cierre

### Lo que debe quedarte de este capítulo

- La familia lineal es la primera **gramática seria del modelado**.
- En regresión lineal, el modelo produce una salida continua y se ajusta minimizando MSE.
- En regresión logística, el modelo produce una probabilidad y se ajusta minimizando log loss.
- La regularización no es un detalle menor: controla complejidad y estabilidad.
- El preprocessing puede afectar:
  - el desempeño,
  - la estabilidad,
  - y la interpretación.
- Una probabilidad estimada no es todavía una acción.
- El threshold es una política y debe justificarse con el costo del error.
